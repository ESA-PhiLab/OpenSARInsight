# Data Preprocessing Tutorial

This notebook guides you through the main data preprocessing steps for the YOLO-Pose vessel detection pipeline.

You will:
- Generate PNG images from raw data
- Generate YOLO-format labels
- (Optional) Filter the dataset
- Extract patch-level metadata for analysis

---

## 1. Generate PNG Images

Use `generate_png.py` to convert raw SAR data into PNG images for training and validation.

- Input: Raw SLC or RC files
- Output: PNG images in the specified directory

**Example usage:**
```bash
python -m large_model.scripts.generate_png_from_slc_tiff --input_dir /path/to/raw --output_dir /path/to/pngs
```

## 2. Generate YOLO Labels

Use `generate_yolo_labels.py` to create YOLO-format label files for each image.

- Input: Source label files (XML, CSV, or other)
- Output: YOLO label files in the target directory

**Example usage:**
```bash
python -m large_model.scripts.generate_yolo_labels_slc --input_dir /path/to/labels --output_dir /path/to/yolo_labels
```

## 3. (Optional) Filter Dataset

Use `filter_dataset.py` to filter images and labels based on custom criteria (e.g., removing low-quality or incomplete samples).

- Input: Image and label directories
- Output: Filtered images and labels

**Example usage:**
```bash
python -m large_model.scripts.filter_dataset --input_dir /path/to/yolo_labels --output_dir /path/to/filtered
```

## 4. Extract Patch Metadata

Use `extract_patch_metadata.py` to generate a CSV summary of patch-level metadata for downstream analysis and bucketed evaluation.

- Input: Image/label directories and metadata sources
- Output: `patch_vessel_metadata.csv`

**Example usage:**
```bash
python -m large_model.scripts.extract_patch_metadata --input_dir /path/to/filtered --output_csv patch_vessel_metadata.csv
```

---

**Tip:** You can run these steps individually from the terminal, or adapt the commands above to your data locations and requirements. For more details, see the script docstrings or use the `--help` flag.
